```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("House_Pricing.csv")
df.head()
```

```python
#removing duplicate rows
df = df.drop_duplicates()
print(df.shape)
```

```python
#drpping duplicate column
duplicate_columns = []
for i in range(df.shape[1]):
    col1 = df.iloc[:, i]
    for j in range(i + 1, df.shape[1]):
        col2 = df.iloc[:, j]
        if col1.equals(col2):
            duplicate_columns.append(df.columns[j])

duplicate_columns
df = df.drop(columns=duplicate_columns)
df.shape
```

```python
#miissing values count
df.isnull().sum()
```

```python
#categorical columns
categorical_cols = df.select_dtypes(exclude=np.number).columns
categorical_cols
```

```python
#numerical cols
numerical_cols = df.select_dtypes(include=np.number).columns
numerical_cols
```

```python
#filling numerical values
for col in numerical_cols:
    df[col].fillna(df[col].mean(), inplace=True)
```

```python
#filling categorical values
for col in categorical_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```


Missing Values
Numerical columns were imputed using the median (or mean) because it preserves the distribution and is less affected by outliers.
Categorical columns were filled using the mode, which replaces missing values with the most frequent category.

```python
df.isnull().sum()
```

```python
#scaling numerical variables using standard scaler
scaler = StandardScaler()
df[numerical_cols] = scaler.fit_transform(df[numerical_cols])
df.head()
```

```python
#encoding categorical variables usin one hot encoding

df = pd.get_dummies(df,
                    columns=categorical_cols,
                    drop_first=True)
df.head()
```

```python
numeric = df.select_dtypes(include=np.number)
numeric.columns
```

```python
#removing outliers
for col in numeric.columns:

    if col == "Sale Price":
        continue
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df = df[(df[col] >= lower) & (df[col] <= upper)]
df.shape
```

```python
#train test split
X = df.drop("Sale Price", axis=1)
y = df["Sale Price"]
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42)
```
